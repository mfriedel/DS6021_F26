```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import os
os.chdir('/Users/lior/DS6021/DS6021_F26/data')

penguins = pd.read_csv('penguins.csv')
peng_subset = penguins.dropna(subset=['bill_depth_mm', 'body_mass_g'])
```

```python
#part a
sns.regplot(data=peng_subset, x="bill_depth_mm", y="body_mass_g", ci=None,
            scatter_kws={'alpha': 0.6, 's': 30}, line_kws={'color': 'green'})
plt.title("Body Mass vs. Bill Depth")
plt.xlabel("Bill Depth (mm)")
plt.ylabel("Body Mass (g)")
plt.show()
```

part a response: the relationsip of the body mass to bill depth is negative which means that as bill depth decreases, the body mass also decreases. The dots are scattered all around, so I would say that the strength of the relationship is moderately weak. I thought this was kind of expected since bill depth and body mass seem to go hand in hand proportion wise.

```python
#part b

x = peng_subset['bill_depth_mm']
y = peng_subset['body_mass_g']
x_bar, y_bar = x.mean(), y.mean()

b1 = ((x - x_bar) * (y - y_bar)).sum() / ((x - x_bar) ** 2).sum()
b0 = y_bar - b1 * x_bar

print(f"body_mass_g = {b0:.2f} + {b1:.2f} * bill_depth_mm")
```

for each bill depth mm, the predicted mody mass decreases by 191.64 which is a linear regression indicator.

```python
#part c

n = len(peng_subset)
y_hat = b0 + b1 * x
residuals = y - y_hat

RSS = (residuals ** 2).sum()
RSE = np.sqrt(RSS / (n - 2))
print(f"RSE = {RSE:.2f}")

r = np.corrcoef(x, y)[0, 1]
R2 = r ** 2
print(f"correlation r = {r:.4f}")
print(f"R^2 = {R2:.4f}")
```

The RSE shows that the prediction of the actual body mass is off by 708.08 grams and R2 shows that the amount of body mass thats explained by bill depth is about 22.27%. This shows that bill depth isnt an accurate predictor of body mass.
